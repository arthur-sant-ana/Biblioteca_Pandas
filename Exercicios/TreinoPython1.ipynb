```python
import json
```

Atividade DNC Input Tarefas

```python
tarefas = {}

while True:
  print("** Menu de opcoes **")
  print("1. Adicionar tarefa")
  print("2. Listar tarefas")
  print("3. Remover tarefa")
  print("4. Carregar tarefas")
  print("5. Salvar tarefas")
  print("6. Sair")

  opcao = input("Selecione:")

  if opcao == '1':
    adicionar = input("Tarefa a ser adicionada: ")
    prioridade = input("prioridade da tarefa: ")
    tarefas[adicionar] = prioridade

  elif opcao == '2':
    for i in tarefas:
      print(f'- {i}, prioridade {tarefas[i]}')

  elif opcao == '3':
    excluir = input("Tarefa a ser excluida: ")
    tarefas.pop(excluir)

  elif opcao == '4':
    with open('tarefas.json', 'r') as arquivo:
      tarefas = json.load(arquivo)
    print('Tarefas carregadas')

  elif opcao == '5':
    with open("tarefas.json", "w") as arquivo:
      json.dump(tarefas, arquivo)
    print('Tarefas salvas')

  elif opcao == '6':
    print('Saindo...')
    break





```
